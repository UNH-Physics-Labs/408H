## Objectives:
- Visualize position, velocity, and acceleration for simple harmonic motion.
- Explore the effect of parameters: amplitude (A), angular frequency (ω), and phase (φ).
- Derive velocity and acceleration from position using calculus.
- Investigate phase shifts and their impact on motion.


```
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
import sympy as sp

%matplotlib inline
```

```
# Define symbolic variables
A, omega, phi, t = sp.symbols('A omega phi t')

# Position function
x = A * sp.sin(omega * t + phi)

# Velocity and acceleration
v = sp.diff(x, t)
a = sp.diff(v, t)

x, v, a
```

```
print(f"Position: x(t) = {x}")
print(f"Velocity: v(t) = {v}")
print(f"Acceleration: a(t) = {a}")
```

```
def motion(A=1.0, omega=2*np.pi, phi=0.0):
    t_vals = np.linspace(0, 2, 500)
    x_vals = A * np.sin(omega * t_vals + phi)
    v_vals = A * omega * np.cos(omega * t_vals + phi)
    a_vals = -A * omega**2 * np.sin(omega * t_vals + phi)

    plt.figure(figsize=(10,6))
    plt.plot(t_vals, x_vals, label='Position x(t)')
    plt.plot(t_vals, v_vals, label='Velocity v(t)')
    plt.plot(t_vals, a_vals, label='Acceleration a(t)')
    plt.xlabel('Time (s)')
    plt.ylabel('Value')
    plt.title('Motion of Object on a Spring')
    plt.legend()
    plt.grid(True)
    plt.show()

interact(motion, A=(0.1, 3.0, 0.1), omega=(1.0, 10.0, 0.5), phi=(0.0, 2*np.pi, 0.1))
```

## Parameter Investigation
- **Amplitude (A):** Controls maximum displacement.
- **Angular Frequency (ω):** Related to period: T = 2π / ω.
- **Phase (φ):** Shifts the motion in time.
