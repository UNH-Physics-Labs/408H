### Scipy

A common way to solve differential equations in Python is using the Scipy library. The way it does this is by taking a starting condition and calculating how the *state* of the system would evolve according to your differential equation in some very small but discrete time increment $dt$. Then, you take the next state and use it as your new "initial" condition. For this purpose, Scipy defines the function `solve_ivp(function, timespan, initial_conditions)`, where `ivp` stands for initial value problem.
