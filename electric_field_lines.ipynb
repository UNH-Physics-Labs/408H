## An Algorithm for Plotting Electric Field Lines

Our goal is to instruct the computer how to construct a plot showing electric field lines. In computing, we call those instructions an *algorithm*, which is a rigid set of steps to be carried out by the computer.

Often, the best way to start with a coding problem is think about what steps you would take if you were doing what you want the computer to do.
To rigorously plot the electric lines, the simplest way is:

- Pick a point
- Use our knowledge of physics to compute what the electric field vector is at that point
- Draw the vector on your plot

If you repeat this for enough points, you get a picture of what the overall field looks like.

For simplicity, we will place a single point charge of magnitude $Q$ (you'll get to pick) at the origin of our coordinate system, $(0, 0)$.

We are going to ask the computer to carry out that process. The first step is to create a grid of points at which the computer will find the vectors.

To create a grid, the easiest way is to construct an *array* of $x$-points and $y$-points, and then combine them into a matrix using the `meshgrid` NumPy function.

### Creating Grids of Points

```python
# Define how many points will be in our grid
num_x_points = 3
num_y_points = 3

# To create an array of linearly spaced points in some range, we use the NumPy function `np.linspace`
# np.linspace(start, stop, num_points) returns an array of length `num_points` with values evenly spaced between start and stop
x_arr = np.linspace(0, 1, num_x_points)
y_arr = np.linspace(1, 3, num_y_points)
# x_arr = np.array([1, 2, 3])
# y_arr = np.array([10, 20, 30])

print(f"The array of x points is: {x_arr}")
print(f"The array of y points is: {x_arr}")
```

```python
# To create a rectangular grid, we need every combination of the x and y points
# This is done with the np.meshgrid(x, y) function, which returns matrices with the 
# x and y coordinates of every point in the grid
X, Y = np.meshgrid(x_arr, y_arr)
print(f"The grid of x-values is:\n {X}") # note the \n stands for "newline"
print(f"The grid of y-values is:\n {Y}")
```

```python
# It helps to visualize the differences between the two, which we do in this cell. 
# **Feel free to gloss over the details here.**
# Create a figure with two subplots (called axes):
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5)) # ax stands for "axis" and refers to a distinct diagram in our overall figure

# The left diagram (ax1) will show the arrays
# scatter takes an array of x points and an array of y points
ax1.scatter(x_arr, np.zeros_like(x_arr), color='red', label='x_arr (1D)') # np.zeros_like creates an array of the same size and shape as input, but all 0
ax1.scatter(np.zeros_like(y_arr), y_arr, color='blue', label='y_arr (1D)')
ax1.set_title("Input Arrays")
ax1.grid(True, linestyle='--') # Define the grid type
ax1.legend() # Plot the legend in the top right of the diagram

# The right diagram (ax2) shows the full meshgrid
ax2.scatter(X, Y, color='purple', marker='x') # notice how easy it is to make this graph?
ax2.set_title("Meshgrid (Full Coordinate Table)")
ax2.grid(True, linestyle='--')

# Display our plot with a little formatting
plt.tight_layout()
plt.show()
```

### Evaluating the Electric Field on our Grid
Now, we have a grid. How do we evaluate the electric field at each point? A convenient way to do this is to define a function `calculate_field` that will output the electric field vector at a point.
What does our function need as inputs to output the electric field? Let's look at the equation to find out:
$$
\vec{E} = \frac{kq}{r^2} \hat{r}
$$

So, we need $q$ and $\vec{r}$. We define $\vec{r}$ as the vector from our point charge to the evaluation point. So that means we need to input:

1. The source charge magnitude $q$
2. The location of the source charge $q_\text{pos} = (x_0, y_0)$
3. The location of our evaluation $(X, Y)$

For this activity, we will only have a single source charge at the origin, so we can "hardcode" the location of the source charge. A convenient way to do that in code is to make it an optional argument to a function.

A nice feature of NumPy is that we can treat grids just as we would single values in what follows.
