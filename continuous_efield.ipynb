Run the following cell to create an interactive plot analyzing your formula.

```python
def visualize_riemann_convergence(N=10):
    """
    Shows how Riemann sum converges to analytical result
    Bottom plot shows E(d) for both numerical and analytical as N changes
    """
    
    d = 0.09  # Distance from P to left end (9 cm)
    
    # Calculate Riemann sum for current N
    dx = L / N # step size
    centers = d + (np.arange(N) + 0.5) * dx # center of each chunk
    E_sum = np.sum(k_e * (Q/N) / centers**2) # sum all the chunks
    E_analytical = part1_solution(d) # analytic solution
    
    percent_error = abs(E_sum - E_analytical) / E_analytical * 100
    
    # Create figure
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 9))
    
    # TOP PLOT: Rod Geometry
    ax1.set_xlim(-0.02, 0.38)
    ax1.set_ylim(-0.05, 0.05)
    ax1.set_aspect('equal')
    
    # Point P
    ax1.plot(0, 0, 'ro', markersize=12)
    ax1.text(0, -0.015, 'P', fontsize=14, fontweight='bold', ha='center')
    
    # Draw rod slices
    colors = ['cornflowerblue', 'lightsteelblue'] * N # Alternating blues
    ax1.bar(centers, 0.04, width=dx, bottom=-0.02, color=colors[:N], edgecolor='darkblue')
    
    # Dimension labels
    ax1.annotate('', xy=(d, -0.03), xytext=(0, -0.03),
                arrowprops=dict(arrowstyle='<->', color='green', lw=2))
    ax1.text(d/2, -0.04, f'd = {d*100:.0f} cm', ha='center', fontsize=11, color='green')
    
    ax1.annotate('', xy=(d+L, 0.03), xytext=(d, 0.03),
                arrowprops=dict(arrowstyle='<->', color='purple', lw=2))
    ax1.text(d+L/2, 0.04, f'L = {L*100:.0f} cm (N={N})', ha='center', fontsize=11, color='purple')

    # Misc formatting
    ax1.set_xlabel('Position (m)', fontsize=12)
    ax1.set_title(f'Rod Divided into N = {N} Slices', fontsize=13, fontweight='bold')
    ax1.set_yticks([])
    ax1.grid(True, alpha=0.3)
    
    # BOTTOM PLOT: E-field vs Distance d
    # Create array of test distances
    d_range = np.linspace(min(0.01, d), 0.30, 50) # 50 points
    
    # Calculate E-field for each distance using both methods
    # Analytical curve
    ax2.plot(d_range*100, part1_solution(d_range)/1e6, 'r-', lw=3, label="Analytical Solution", alpha=0.6)
    
    # Numerical curve for current N
    E_num_range = [np.sum(k_e * (Q/N) / (d + (np.arange(N)+0.5)*(L/N))**2) for d in d_range]
    ax2.plot(d_range*100, np.array(E_num_range)/1e6, 'b--', label=f"Numerical (N={N})")
    
    # Current evaluation point
    ax2.plot(d*100, E_sum/1e6, 'go', markersize=10, label="Current Result")
    
    ax2.set_ylabel("E-Field Strength ($10^6$ N/C)")
    ax2.set_xlabel("Distance $d$ from Point P (cm)")
    ax2.legend()
    ax2.grid(True, alpha=0.2)

    plt.tight_layout()
    plt.show()
    
    # Print results below plot
    print(f"Numerical Result:  {E_sum:.3e} N/C")
    print(f"Analytical Result: {E_analytical:.3e} N/C")
    print(f"Error: {abs(E_sum - E_analytical)/E_analytical*100:.4f}%")

# Interactive widget
interact(visualize_riemann_convergence, 
         N=IntSlider(min=1, max=50, step=1, value=1, description='N slices:'));
```

```python
def visualize_perpendicular_case(N=8, h=0.12):
    # --- 1. COMPUTATION ---
    dx = L / N
    x_centers = np.linspace(-L/2 + dx/2, L/2 - dx/2, N)
    dq = Q / N
    
    r = np.sqrt(x_centers**2 + h**2)
    cos_theta = h / r
    sin_theta = x_centers / r # Needed for horizontal component
    
    # Components of the field from each slice
    dEy = (k_e * dq / r**2) * cos_theta
    dEx = (k_e * dq / r**2) * (-sin_theta) # Points opposite to the x-position
    
    # Net fields (E_x should sum to ~0)
    E_sum_x = np.sum(dEx)
    E_sum_y = np.sum(dEy)
    E_ana = part2_solution(h)

    # Now we can start plotting!
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # LEFT PLOT: Geometry & Vectors
    ax1.set_title(f"Field Vectors at Point P (h={h*100:.0f} cm)")
    ax1.plot(0, h, 'ro', markersize=10, label="Point P")
    
    colors = ['cornflowerblue', 'lightsteelblue'] * N
    ax1.bar(x_centers, 0.015, width=dx, bottom=-0.0075, color=colors[:N], edgecolor='darkblue')
    
    # Draw representative TOTAL dE vectors
    scale = 0.05 / np.max(dEy) 
    for i in range(0, N, max(1, N//8)):
        # Draw the full vector (dEx and dEy)
        ax1.arrow(0, h, dEx[i]*scale, dEy[i]*scale, width=0.003, color='purple', alpha=0.6)
        ax1.plot([x_centers[i], 0], [0, h], 'gray', lw=1, ls=':', alpha=0.2)

    ax1.set_xlim(-0.2, 0.2); ax1.set_ylim(-0.05, 0.3)
    ax1.set_aspect('equal')
    ax1.legend()

    # RIGHT PLOT: Symmetry Visualization
    ax2.set_title(f"Net $E_x$: {E_sum_x:.1e} | Net $E_y$: {E_sum_y:.2e}")
    
    # Plot dEy (Symmetric) and dEx (Anti-symmetric)
    ax2.bar(x_centers*100 - 0.2, dEy/1e6, width=dx*40, color='cornflowerblue', label="$dE_y$ (Vertical)")
    ax2.bar(x_centers*100 + 0.2, dEx/1e6, width=dx*40, color='salmon', label="$dE_x$ (Horizontal)")
    
    ax2.axhline(0, color='black', lw=1)
    ax2.set_xlabel("Slice Position (cm)")
    ax2.set_ylabel("Contribution ($10^6$ N/C)")
    ax2.legend()
    ax2.grid(True, alpha=0.2)

    plt.tight_layout()
    plt.show()

interact(visualize_perpendicular_case, 
         N=IntSlider(min=2, max=60, step=2, value=2),
         h=FloatSlider(min=0.05, max=0.3, step=0.01, value=0.12));
```

## Bonus: Visualizing Field Lines

No questions are related to the following, but you may find it satisfying to explore.

```python
def visualize_field_lines(N=20, h_point=0.12):
    # Define our step sizes
    dx = L / N
    dq = Q / N
    x_slices = np.linspace(-L/2 + dx/2, L/2 - dx/2, N)

    # Create a grid
    x_grid = np.linspace(-0.35, 0.35, 40)
    y_grid = np.linspace(-0.25, 0.35, 40)
    X, Y = np.meshgrid(x_grid, y_grid)
    Ex, Ey = np.zeros_like(X), np.zeros_like(Y)
    
    # Compute the field
    for xs in x_slices:
        rx, ry = X - xs, Y - 0
        r = np.sqrt(rx**2 + ry**2)
        
        # dE = k * dq / r^2 in the direction of r_vector / r
        Ex += k_e * dq * rx / r**3
        Ey += k_e * dq * ry / r**3
    
    # Plotting
    fig, ax = plt.subplots(figsize=(10, 8))
    
    # Magnitude for coloring (using log scale so detail is visible)
    mag = np.sqrt(Ex**2 + Ey**2)
    
    strm = ax.streamplot(X, Y, Ex, Ey, color=np.log10(mag), 
                          cmap='plasma', linewidth=1.2, density=1.5)
    
    plt.colorbar(strm.lines, label='Field Strength log10(E)')
    
    # Draw the Rod Chunks
    colors = ['red', 'salmon'] * N
    ax.bar(x_slices, 0.02, width=dx, bottom=-0.01, color=colors[:N], edgecolor='black', zorder=5)
    
    # Label Point P and Analytical Result
    E_at_P = (k_e * Q) / (h_point * np.sqrt((L/2)**2 + h_point**2))
    ax.plot(0, h_point, 'go', markersize=10, label=f"Point P (h={h_point:.3f}m)")
    
    ax.set_title(f"Field Lines: {N} Discrete Point Charges\nE at P (Analytical): {E_at_P:.2e} N/C", fontsize=14)
    ax.set_xlabel("x (m)")
    ax.set_ylabel("y (m)")
    ax.legend(loc='upper right')
    ax.grid(True, alpha=0.1)
    
    plt.tight_layout()
    plt.show()

interact(visualize_field_lines,
         N=IntSlider(min=2, max=60, step=2, value=2, description='N slices:'),
         h_point=FloatSlider(min=0.05, max=0.30, step=0.01, value=0.12, description='Height h:'));
```
