# Full Interactive Notebook: Understanding the sign in the phase $kx \pm \omega t$
This notebook is based on the activity *Understanding the sign in the phase $kx \pm \omega t$*.

## Objectives:
- Visualize a traveling wave
- Explore how the phase effects the movement of the wave

```python
import micropip
import asyncio
await asyncio.gather(
    micropip.install("ipywidgets"),
    micropip.install("numpy"),
    micropip.install("ipympl"),
    micropip.install("matplotlib")
)
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, ToggleButtons, Layout

%matplotlib inline
```

```python
# Physical constants
k = 2.0 * np.pi      # Wavenumber
omega = 2.0 * np.pi  # Angular Frequency
v = omega / k        # Phase velocity (1 m/s)

# State Variables
# We use these to "remember" where the wave was, so it doesn't jump
# when you switch directions.
current_offset = 0.5          
previous_direction = 'Right (Minus)'
```

```python
def plot_wave(direction, t):
    # We need to update the external state, so we use the 'global' keyword
    global current_offset, previous_direction
    
    # Check if the direction button was just switched
    if direction != previous_direction:
        # To make the motion continuous, we calculate where the crest 
        # is right now and force the new wave to start there.
        
        # Calculate the crest's position using the OLD direction
        old_sign = 1 if previous_direction == 'Right (Minus)' else -1
        current_pos = (old_sign * v * t) + current_offset
        
        # Calculate the NEW offset required to keep the crest at that position
        new_sign = 1 if direction == 'Right (Minus)' else -1
        current_offset = current_pos - (new_sign * v * t)
        
        # Update the 'memory' variable
        previous_direction = direction

    # Calculate the current position of the crest
    sign = 1 if direction == 'Right (Minus)' else -1
    x_crest = (sign * v * t) + current_offset
    
    # Generate the Wave Data
    # We calculate the wave relative to the crest position (x_crest)
    x_vals = np.linspace(0, 4, 500)
    y_vals = 1.0 * np.cos(k * (x_vals - x_crest))
    
    # Determine the math symbol for the title
    sign_symbol = '-' if direction == 'Right (Minus)' else '+'

    # Plot the Graph
    plt.figure(figsize=(10, 5))
    plt.plot(x_vals, y_vals, linewidth=2, color='#0077BE', label='Wave')
    
    # Draw the Red Dot (Crest Tracker) if it is visible on screen
    if 0 <= x_crest <= 4:
        plt.plot(x_crest, 1.0, 'ro', markersize=10, label='Tracked Crest')
        
    plt.title(fr"Wave Phase: $kx {sign_symbol} \omega t$", fontsize=14)
    plt.xlabel("Position x (m)")
    plt.ylabel("Displacement y (m)")
    plt.ylim(-1.5, 1.5)
    plt.grid(True, alpha=0.3)
    plt.legend(loc='upper right')
    plt.show()
```

```python
# Create the user interface elements
# Create the toggle button for Direction
dir_widget = ToggleButtons(
    options=['Right (Minus)', 'Left (Plus)'],
    description='Direction:',
    button_style='info'
)

# Create the slider for Time
time_widget = FloatSlider(
    value=0.0, min=0.0, max=4.0, step=0.05,
    description='Time (s):', layout=Layout(width='60%')
)
```

```python
# Launch the interactive plot!
print("Interact with the simulation below:")
interact(plot_wave, direction=dir_widget, t=time_widget);
```
